```python
# ============================================================
# Notebook 08
# Sensor Confidence Index (SCI)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

plt.rcParams["figure.figsize"]=(10,6)
plt.rcParams["figure.dpi"]=120
```

```python
# ============================================================
# Cell 2
# Load Sensor Quality Metrics
# ============================================================

PROJECT_ROOT = Path(
    r"D:\Research\Papers\Paper_2"
)

RESULT_DIR = PROJECT_ROOT / "results"

FIGURE_DIR = PROJECT_ROOT / "figures"

DRIVE = "2011_09_26_drive_0001_sync"

quality = pd.read_csv(

    RESULT_DIR /

    f"sensor_quality_metrics_{DRIVE}.csv"

)

print("=" * 70)

print("Sensor Quality Metrics Loaded")

print("=" * 70)

print("Drive :", DRIVE)

print("Frames :", len(quality))

print("=" * 70)

display(quality.head())
```

```python
metrics = [

"GPS Position Stability",

"GPS Velocity Stability",

"GPS Heading Stability",

"Satellite Quality",

"Position Accuracy Score",

"Velocity Accuracy Score",

"Motion Smoothness",

"Angular Stability"

]

display(
    quality[metrics].describe()
)
```

```python
gps_weights = {

"GPS Position Stability":0.20,

"GPS Velocity Stability":0.20,

"GPS Heading Stability":0.15,

"Satellite Quality":0.15,

"Position Accuracy Score":0.15,

"Velocity Accuracy Score":0.15

}

quality["GPS Confidence"]=(

      gps_weights["GPS Position Stability"]*quality["GPS Position Stability"]

    + gps_weights["GPS Velocity Stability"]*quality["GPS Velocity Stability"]

    + gps_weights["GPS Heading Stability"]*quality["GPS Heading Stability"]

    + gps_weights["Satellite Quality"]*quality["Satellite Quality"]

    + gps_weights["Position Accuracy Score"]*quality["Position Accuracy Score"]

    + gps_weights["Velocity Accuracy Score"]*quality["Velocity Accuracy Score"]

)

display(

quality[

["GPS Confidence"]

].head()

)
```

```python
imu_weights={

"Motion Smoothness":0.50,

"Angular Stability":0.50

}

quality["IMU Confidence"]=(

      imu_weights["Motion Smoothness"]*quality["Motion Smoothness"]

    + imu_weights["Angular Stability"]*quality["Angular Stability"]

)

display(

quality[

["IMU Confidence"]

].head()

)
```

```python
quality["Sensor Confidence Index"]=(

      0.60*quality["GPS Confidence"]

    + 0.40*quality["IMU Confidence"]

)

display(

quality[[

"GPS Confidence",

"IMU Confidence",

"Sensor Confidence Index"

]].head()

)
```

```python
# ============================================================
# Confidence Classification
# ============================================================

def classify(score):

    if score >= 0.90:
        return "Excellent"

    elif score >= 0.75:
        return "High"

    elif score >= 0.60:
        return "Moderate"

    elif score >= 0.40:
        return "Low"

    else:
        return "Poor"


quality["Confidence Level"] = (

    quality["Sensor Confidence Index"]

    .apply(classify)

)

display(

quality[
[
"Sensor Confidence Index",
"Confidence Level"
]
].head()
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(

    quality["Sensor Confidence Index"],

    bins=15,

    edgecolor="black"

)

plt.xlabel("Sensor Confidence")

plt.ylabel("Frequency")

plt.title("Sensor Confidence Distribution")

plt.grid(True)

plt.tight_layout()

plt.savefig(

    FIGURE_DIR /
    "sensor_confidence_distribution.png",

    dpi=300

)

plt.show()
```

```python
# ============================================================
# Confidence Statistics
# ============================================================

statistics = quality[[

    "GPS Confidence",

    "IMU Confidence",

    "Sensor Confidence Index"

]].describe()

display(statistics)

summary = pd.DataFrame({

    "Metric":[

        "Average GPS Confidence",

        "Average IMU Confidence",

        "Average Sensor Confidence",

        "Maximum SCI",

        "Minimum SCI",

        "Standard Deviation"

    ],

    "Value":[

        quality["GPS Confidence"].mean(),

        quality["IMU Confidence"].mean(),

        quality["Sensor Confidence Index"].mean(),

        quality["Sensor Confidence Index"].max(),

        quality["Sensor Confidence Index"].min(),

        quality["Sensor Confidence Index"].std()

    ]

})

display(summary)

statistics.to_csv(

    RESULT_DIR/

    "sensor_confidence_statistics.csv"

)

summary.to_csv(

    RESULT_DIR/

    "sensor_confidence_summary.csv",

    index=False

)
```

```python
# ============================================================
# Correlation Matrix
# ============================================================

correlation = quality[[

    "GPS Confidence",

    "IMU Confidence",

    "Sensor Confidence Index"

]].corr()

display(correlation)

fig, ax = plt.subplots(figsize=(6,5))

image = ax.imshow(

    correlation,

    cmap="viridis"

)

for i in range(len(correlation.columns)):

    for j in range(len(correlation.columns)):

        ax.text(

            j,

            i,

            f"{correlation.iloc[i,j]:.2f}",

            ha="center",

            va="center",

            color="white",

            fontsize=11

        )

ax.set_xticks(range(len(correlation.columns)))

ax.set_xticklabels(

    correlation.columns,

    rotation=30

)

ax.set_yticks(range(len(correlation.columns)))

ax.set_yticklabels(

    correlation.columns

)

plt.colorbar(image)

plt.title("Correlation Matrix of Sensor Confidence")

plt.tight_layout()

plt.savefig(

    FIGURE_DIR/

    "confidence_correlation_matrix.png",

    dpi=300

)

plt.show()
```

```python
# ============================================================
# Cell 11
# Save Results
# ============================================================

quality.to_csv(

    RESULT_DIR /

    f"sensor_confidence_index_{DRIVE}.csv",

    index=False

)

statistics.to_csv(

    RESULT_DIR /

    f"sensor_confidence_statistics_{DRIVE}.csv"

)

summary.to_csv(

    RESULT_DIR /

    f"sensor_confidence_summary_{DRIVE}.csv",

    index=False

)

correlation.to_csv(

    RESULT_DIR /

    f"sensor_confidence_correlation_{DRIVE}.csv"

)

print("=" * 70)

print("Sensor Confidence Results Saved Successfully")

print("=" * 70)
```
